```python
from gensim.models import Word2Vec
import numpy as np
```

```python
sentences = [
    ["the", "cat", "sat"],
    ["the", "dog", "ran"],
    ["the", "cat", "ran"],
    ["the", "dog", "sat"]
]
```

```python
# Train Word2Vec
model = Word2Vec(
    sentences=sentences,
    vector_size=4,
    window=2,
    min_count=1,
    workers=1
)
```

```python
# Sequence we want to encode
sequence = ["the", "cat", "sat"]
```

```python
# Convert every word into its Word2Vec vector
encoded_sequence = np.array([
    model.wv[word] for word in sequence
])
```

```python
print(encoded_sequence)
print(encoded_sequence.shape)
```
